```python
# exemplo de np.array
example_array = np.array([10, 20, 30])
print("np.array exemplo:", example_array)

# exemplo de type()
print("type(array1):", type(array1))

# exemplo de .ndim
print("array1.ndim:", array1.ndim)
print("array2.ndim:", array2.ndim)

# exemplo de np.zeros
example_zeros = np.zeros((2, 2))
print("np.zeros exemplo:\n", example_zeros)

# exemplo de np.ones
example_ones = np.ones(4)
print("np.ones exemplo:", example_ones)
```

```python
# Arithmetic operations
print("Addition:", array1 + 5)
print("Subtraction:", array1 - 2)
print("Multiplication:", array1 * 3)
print("Division:", array1 / 2)
print("Power:", array1 ** 2)

# Array operations
print("Sum:", array1.sum())
print("Mean:", array1.mean())
print("Min:", array1.min())
print("Max:", array1.max())
print("Std Dev:", array1.std())

# Element-wise operations between arrays
print("Element-wise add:", array1 + np.array([5, 10, 15, 20, 25]))
print("Element-wise multiply:", array1 * np.array([2, 2, 2, 2, 2]))

# Matrix operations with array2
print("array2 sum:", array2.sum())
print("array2 sum by axis 0:", array2.sum(axis=0))
print("array2 sum by axis 1:", array2.sum(axis=1))

# Matrix multiplication
print("Matrix multiply:", np.dot(array2, array2.T))

# Reshaping
print("Reshape array1:", array1.reshape(5, 1))

# Concatenation
print("Concatenate:", np.concatenate([array1, np.array([6, 7])]))

# Sorting
print("Sort:", np.sort(array1[::-1]))
```
